```python
import json
from pathlib import Path
from html import escape
from IPython.display import HTML, display
root = Path.cwd().resolve()
if root.name == 'notebooks':
    root = root.parent
result_dir = root / 'outputs/style_aware_steps_2026-10-09/classified'
summary = json.loads((result_dir / 'summary.json').read_text(encoding='utf-8'))
print({k: v for k, v in summary.items() if k != 'rows'})
names = {'continuous_approach_candidate': '连续助跑候选', 'terminal_adjustment_candidate': '末段调整候选', 'undetermined': '暂不能判断'}
rows = []
for r in summary['rows']:
    values = [r['clip_id'], names[r['attempt_style']], r['candidate_step_count'], r['count_partition']['before_terminal'], r['count_partition']['terminal_adjustment'], r['count_partition']['contact_guard'], r['count_sensitivity']['threshold_and_contact_range']]
    rows.append('<tr>' + ''.join('<td>' + escape('—' if v is None else str(v)) + '</td>' for v in values) + '</tr>')
display(HTML('<table><thead><tr><th>视频</th><th>方式</th><th>总候选</th><th>末段前</th><th>末段</th><th>触球保护区</th><th>参数范围</th></tr></thead><tbody>' + ''.join(rows) + '</tbody></table>'))
```
